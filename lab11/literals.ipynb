### Summary of Learning
- **Literals:** Fixed values in code like numbers (`321`, `2.34`), strings (`"Hello"`), and booleans (`True`, `False`).
- **Number systems:** Binary (base 2), Octal (base 8), Hexadecimal (base 16).
- **Escape characters:** Using `\` to add special characters like quotes (`\'`) or new lines (`\n`).
- **print():** Takes values (arguments) and formatting options (keyword arguments like `sep` and `end`).

```python
print(1 > 2)
print(1 == 2)
print(2 > 1)
print("A" == "A")
print(True > False)
print('I\'m happy.')
```
