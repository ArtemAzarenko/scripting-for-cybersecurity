```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number.")
else:
    print("Odd number.")
```

```python
ch = input("Enter a char: ")

if ch == 'a':
    print("Apple")
elif ch == 'b':
    print("Ball")
else:
    print("Other Chars")
```

```python
charset = ['f', 'a', 'b', 'c', 'j', 'i', 'e', '*', ')', '£', '2', '/', 's', 'o']

# 1. Using a for loop
found_for = False
for char in charset:
    if char == 'e':
        found_for = True
        break

if found_for:
    print("for loop: 'e' is in the list.")
else:
    print("for loop: 'e' is not in the list.")

# 2. Using a while loop
found_while = False
i = 0
while i < len(charset):
    if charset[i] == 'e':
        found_while = True
        break
    i += 1

if found_while:
    print("while loop: 'e' is in the list.")
else:
    print("while loop: 'e' is not in the list.")
```

```python
word = input("Please enter a word: ")
print(list(word))
print(word)

reversed_word = word[::-1]
print(reversed_word)

if word == reversed_word:
    print("palindrome")
else:
    print("not a palindrome")
```

```python
n = int(input("Enter number of elements of the list to be created : "))
numbers = []

for _ in range(n):
    val = int(input())
    numbers.append(val)

first_and_last = [numbers[0], numbers[-1]]
print(first_and_last)
```
